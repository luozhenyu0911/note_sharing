```bash
Installing, this may take a few minutes...
WslRegisterDistribution failed with error: 0x800701bc
Error: 0x800701bc WSL 2 ????????????? https://aka.ms/wsl2kernel
Press any key to continue...
```
    是由于在wsl1升级到wsl2后，没有安装wsl2的内核升级包

# 3.在Ubuntu和VScode中配置jupyter notebook环境
- **1. 在Ubuntu中安装conda，这里我选用miniconda，因为conda安装包较小，速度也快，满足生信环境的需求。**
打开Ubuntu终端，然后输入以下命令下载miniconda安装包：
```bash
# 从清华镜像站下载最新版本（推荐）
wget https://mirrors.tuna.tsinghua.edu.cn/anaconda/archive/Anaconda3-2024.10-1-Linux-x86_64.sh

# 或者从官方源下载
wget https://repo.anaconda.com/archive/Anaconda3-2024.10-1-Linux-x86_64.sh

# 安装anaconda
bash Anaconda3-2024.10-1-Linux-x86_64.sh

# 选择安装路径，默认安装到/home/username/anaconda3
```
这里安装conda，conda是一个包管理器，可以方便地安装、卸载、管理各种包。如果安装conda时，安装路径不是自己满意的，可以卸载重装。

- **2. 安装jupyter notebook**
安装conda后，会有一个base环境并带有python3环境，当然我们可以创建其他环境
```bash
# 激活base环境 
conda activate base
# 或者 创建环境
# conda create --name TheNameYouWant -y
# conda activate TheNameYouWant

```bash
# 安装Jupyter Notebook及其插件
conda install jupyter notebook -y
python -m ipykernel install --user 
conda install -c conda-forge jupyter_contrib_nbextensions -y
conda install jupyterthemes -y

- CondaToSNonInteractiveError: Terms of Service have not been accepted for the following channels. Please accept or remove them before proceeding:
    - https://repo.anaconda.com/pkgs/main
    - https://repo.anaconda.com/pkgs/r
如果遇到此错误，请执行以下命令：

conda config --set always_yes yes
conda config --set auto_update_conda false
```

- **3. 启动jupyter notebook**
```bash
# 启动jupyter notebook
# 直接在Ubuntu终端输入 code . 命令会打开VScode
code .
```
然后在VScode中 按 Ctrl+Shift+P 打开命令面板，输入Jupyter: Create New Blank Notebook 新建一个空白的jupyter notebook。
然后就可以在**选择内核**中选择Unbuntu中的anaconda环境中python解释器，打开jupyter notebook编辑代码了。
