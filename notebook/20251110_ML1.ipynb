# 1. 逻辑回归（Logistic Regression）



**简而言之：逻辑回归是一种分类算法，就是给一堆特征，判断是 0 还是 1（比如垃圾邮件/正常邮件）。把每个特征乘以一个“权重”，加起来后通过 Sigmoid 函数挤成 0~1 的概率。**

## 算法逻辑
1.  输入特征 $x_1$, $x_2$, $\ldots$, $x_n$
2.  计算线性组合：$$ z = w_1 x_1 + w_2 x_2 + \ldots + b $$
3.  过 Sigmoid：$$ p = \frac{1}{1 + e^{-z}} \rightarrow \text{概率} $$
4.  用梯度下降不断调权重 ( w, b ) 让损失最小

# 2. 随机森林（Random Forest）
    

**随机森林是一种集成学习算法，它通过构建多棵决策树，并综合这些树的预测结果（如通过投票或平均）来进行分类或回归任务**

## 算法逻辑

1. 从数据中有放回抽样，造多个子数据集  
2. 对于每个子数据集，随机选一部分特征（防过拟合）  
3. 每棵树独立长大（不剪枝）
4. 预测时：分类-投票法，回归-平均法  


# 3. XGBoost

**XGBoost 是一种高性能的梯度提升算法，它通过集成多棵决策树，并利用损失函数的二阶泰勒展开及正则化等技术，在各类机器学习任务中表现出色。简而言之就是聪明地一个接一个加决策树，每棵树修正前面的错误**

## 算法逻辑：

1. 第一棵树先拟合数据  
2. 计算残差（真实 - 预测）  
3. 第二棵树专门学残差  
4. 总预测 = 树1 + 学习率×树2 + 学习率×树3 + ...  
5. 加正则项（控制树复杂）+ 二阶导数（更快收敛）


# 4. 神经网络（TensorFlow/Keras）


**神经网络是一种模仿人脑神经元网络结构的计算模型，它通过大量节点的相互连接和权重调整，能够从数据中学习并识别复杂的非线性模式。像多层过滤器：第一层看边缘，第二层看形状，最后认出猫。**

## 算法逻辑
1. 输入 → 隐藏层1（加权 + 激活函数）→ 隐藏层2 → ... → 输出
2. 激活函数：ReLU（负数变0）、Sigmoid、Softmax  
3. 损失函数：分类用交叉熵  
4. 反向传播 + Adam 优化器自动调权重
